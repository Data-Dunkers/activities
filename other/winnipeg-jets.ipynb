# Data Dunkers Activity Notebook: Winnipeg Jets Points Leaders

# Objectives

Students will be able to:

- Retrieve data from a CSV file using a URL.
- Process a DataFrame by sorting values.
- Filter or extract a subset of data (top 10).
- Visualize the processed data using a bar chart in Plotly Express.

# Input

In this section, we import the necessary libraries and load the data from the Winnipeg Jets 2025-2026 season CSV file.

```python
import pandas as pd
import plotly.express as px

# URL to the Winnipeg Jets players data
url = 'https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NHL/team_players/2025-2026/WPG_2025-2026_players.csv'

# Load the data into a DataFrame
df = pd.read_csv(url)

# Display the first few rows of the data
display(df.head())
```

# Process

Now we process the data to find the top ten players by points (`PTS`). We will sort the DataFrame in descending order and then take the top 10 rows.

```python
# Sort the players by Points (PTS) in descending order
df_sorted = df.sort_values(by='PTS', ascending=False)

# Select the top ten players
top_ten_df = df_sorted.head(10)

# Display the processed data
display(top_ten_df)
```

# Output

Finally, we use Plotly Express to create a bar chart of the top ten players and their points.

```python
# Create a bar chart showing the top ten players by points
fig = px.bar(top_ten_df, 
             x='Name', 
             y='PTS', 
             title='Winnipeg Jets Top 10 Point Leaders (2025-2026)',
             labels={'Name': 'Player Name', 'PTS': 'Points'})

# Show the plot
fig.show()
```
