# Data Dunkers Activity Notebook: Winnipeg Jets Player Stats Scatter Plot

# Objectives

Students will be able to:

- Load a large dataset from a URL.
- Filter data based on a specific criteria (Team name).
- Create a scatter plot to analyze the relationship between two variables (Goals and Assists).

# Input

In this section, we load the full NHL player statistics for the 2025-2026 season.

```python
import pandas as pd
import plotly.express as px

# URL to the NHL player stats data
url = 'https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NHL/player/nhl_player_stats_2025-2026.csv'

# Load the data
df = pd.read_csv(url)

# Display the first few rows
display(df.head())
```

# Process

We filter the data to include only players on the Winnipeg Jets (`WPG`).

```python
# Filter for Winnipeg Jets players
wpg_df = df[df['Team'] == 'WPG']

# Display the filtered data
display(wpg_df.head())
```

# Output

We create a scatter plot showing Goals (`G`) vs Assists (`A`) for all Winnipeg Jets players.

```python
# Create the scatter plot
fig = px.scatter(wpg_df, 
                 x='G', 
                 y='A', 
                 hover_name='Name', 
                 title='Winnipeg Jets: Goals vs Assists (2025-2026)',
                 labels={'G': 'Goals', 'A': 'Assists'})

# Show the plot
fig.show()
```
